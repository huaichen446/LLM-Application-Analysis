Interactive Visualization: Arxiv Paper Trends and Keyword Evolution

```python
import pandas as pd
import plotly.express as px
from pathlib import Path
```

1️⃣ Path Setup

```python
DATA_PATH = Path("../data/raw/arxiv_trends_by_month.csv")
INTERACTIVE_DIR = Path("../results/interactive")
INTERACTIVE_DIR.mkdir(parents=True, exist_ok=True)

print(f"📂 Loading data file: {DATA_PATH}")
```

2️⃣ Load Data

```python
df = pd.read_csv(DATA_PATH)
df["year"] = df["year"].astype(int)

keyword_cols = [c for c in df.columns if c not in ["year", "total_publications"]]
melted = df.melt(id_vars=["year"], value_vars=keyword_cols,
                 var_name="keyword", value_name="count")
```

3️⃣ Plotly: Overall Paper Trend

```python
fig_total = px.line(
    df,
    x="year",
    y="total_publications",
    markers=True,
    title="📈 LLM-related Publications on Arxiv (2010–2025)",
    labels={"year": "Year", "total_publications": "Total Papers"}
)
fig_total.update_layout(template="plotly_white", hovermode="x unified")
fig_total.write_html(INTERACTIVE_DIR / "arxiv_total_trend.html")
print("Saved: results/interactive/arxiv_total_trend.html")
```

4️⃣ Plotly: Keyword Trends (Top 15)

```python
top_keywords = (
    melted.groupby("keyword")["count"]
    .mean().sort_values(ascending=False).head(15).index.tolist()
)
top_df = melted[melted["keyword"].isin(top_keywords)]

fig_kw = px.line(
    top_df,
    x="year",
    y="count",
    color="keyword",
    markers=True,
    title="Top 15 Keyword Trends in LLM Research",
    labels={"year": "Year", "count": "Paper Count", "keyword": "Keyword"}
)
fig_kw.update_layout(template="plotly_white", hovermode="x unified")
fig_kw.write_html(INTERACTIVE_DIR / "arxiv_keyword_trends.html")
print("Saved: results/interactive/arxiv_keyword_trends.html")
```

5️⃣ Plotly: Keyword Share Trend

```python
for kw in keyword_cols:
    df[f"{kw}_share"] = df[kw] / df["total_publications"] * 100

share_cols = [c for c in df.columns if c.endswith("_share")]
share_melt = df.melt(id_vars=["year"], value_vars=share_cols,
                       var_name="keyword", value_name="share")
share_melt["keyword"] = share_melt["keyword"].str.replace("_share", "")

top_share = (
    share_melt.groupby("keyword")["share"]
    .mean().sort_values(ascending=False).head(15).index.tolist()
)
share_df = share_melt[share_melt["keyword"].isin(top_share)]

fig_share = px.line(
    share_df,
    x="year",
    y="share",
    color="keyword",
    markers=True,
    title="Top 15 Keyword Share in LLM Publications (%)",
    labels={"year": "Year", "share": "Share (%)", "keyword": "Keyword"}
)
fig_share.update_layout(template="plotly_white", hovermode="x unified")
fig_share.write_html(INTERACTIVE_DIR / "arxiv_keyword_share.html")
print("Saved: results/interactive/arxiv_keyword_share.html")
```
